```python
valid_name =[]
while True:
    name = input("Enter name: ")
   # ask a user for input using a while loop
    if name.lower() == "stop":
        print("Goodbye")
        
        #print valid names in the list
        print(f"Valid names: {valid_name}!")
        break

    
    #it should only accept names that contain letters only (no numbers, no symbols, no empty space
    if not name.isalpha() or ' ' in name:
        print("Enter only Letters without space!")
        continue
    #append valid names to empty list    
    valid_name.append(name)  
    print(f"Hello {name}!")
```
