```python
valid_names= []

while True:
    name = input("Enter name: ")
   # ask a user for input using a while loop
    if name.lower() == "stop":
        print("Goodbye")
        break

    
    #it should only accept names that contain letters only (no numbers, no symbols, no empty space
    if not name.isalpha() or ' ' in name:
        print("Enter only Letters without space!")
        continue

        valid_names.append(name)
    print(f"Hello {name}!")    

    
```
