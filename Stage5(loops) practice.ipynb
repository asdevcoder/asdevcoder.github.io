```python
import csv
import pandas as pd
import os #to append rows without headers

valid_name =[]
users= []
while True:
    name = input("Enter name: ").strip()
   # ask a user for input using a while loop
    if name.lower() == "stop":
        print("Goodbye")
        
        #print valid names in the list
        print(f"Valid names: {valid_name}!")
        break

    
    #it should only accept names that contain letters only (no numbers, no symbols, no empty space
    if not name.isalpha() or ' ' in name:
        print("Enter only Letters without space!")
        continue
    #append valid names to empty list    
    valid_name.append(name)  
    print(f"Hello {name}!")
    
    #Enter Age and validate
    theAge= input('Enter age: ').strip()
    try:
        age= int(theAge)
    except ValueError:
        print('Error: Age must be a whole number (e.g., 25).')
        continue

    # Check the age category
    if age <18:
        category =  "Minor"
    elif age >= 18 and age <=  59:
        category= "Adult"
    else:
        category= "Senior citizen"
    
    #Enter color and validate
    colour= input('Enter colour').strip()
    if not colour.isalpha():
        print('Error: Colour must contain letters only')
        continue
        
   
    #create a dictionary and append to the user list

    user= {'name': name, 'age': age, 'colour':colour, 'category':category}
    
    users.append(user)
    print(f"{name} added ✅")

print("\nSummary of users entered:")
    
if not users:
    print("No users were entered.")
else:
    for u in users:
        print(f"- {u['name']} ({u['category']})- {u['age']} years old - Favourite colour: {u['colour']}")

# Save users data into a CSV file
'''
with open('users_data.csv', mode='a', newline='') as file:
    writer = csv.DictWriter(file, fieldnames=['name', 'age', 'colour', 'category'])
    writer.writeheader()  # write the column titles
    writer.writerows(users)  # write all user dictionaries
'''

#Save users data into a CSV file if it exist
file_exists = os.path.exists('users_data.csv')

with open('users_data.csv', mode='a', newline='') as file:
    writer = csv.DictWriter(file, fieldnames=['name', 'age', 'colour', 'category'])
    
    if not file_exists:
        writer.writeheader()  # write the header only once
        
    writer.writerows(users)  # write user data

print("✅ Data has been saved to users_data.csv")

# Read and display the saved data using pandas
data = pd.read_csv('users_data.csv')
print("\n Here's your saved data:")

print(data)
```
